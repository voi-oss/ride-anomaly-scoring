```python
import pandas as pd
```

```python
GBG500 = pd.read_parquet("../data/GBG500.parquet")
SRV3490 = pd.read_parquet("../data/SRV3490.parquet")
```

```python
for name, df in [("GBG500", GBG500), ("SRV3490", SRV3490)]:
    total_hours = df.groupby("ride_id")["time_index"].max().sum() / 3600000
    print(f"{name}: {len(df):,} rows, {total_hours:.1f} hours total")
```

```python
for name, df in [("GBG500", GBG500), ("SRV3490", SRV3490)]:
    duration_min = df.groupby("ride_id")["time_index"].max() / 60000
    duration_min.name = "duration_min"
    print(f"\n{name} — ride duration (minutes):")
    display(duration_min.describe())
```

```python
imu_cols = ["ax", "ay", "az", "rx", "ry", "rz"]

for name, df in [("GBG500", GBG500), ("SRV3490", SRV3490)]:
    missing = df[imu_cols].isna().any(axis=1).groupby(df["ride_id"]).mean()
    missing.name = "frac_missing"
    print(f"\n{name} — fraction of rows with missing values per ride:")
    display(missing.describe())
```
